```python
dbutils.widgets.text(name="env", defaultValue="", label = "environment name please")
env = dbutils.widgets.get("env")
```

```python
def create_bronze_table(env):
    print(f"creating bronze table with in {env}_catalog under bronze schema")
    spark.sql(f"""drop table if exists `{env}_catalog`.`bronze`.`raw_patient`""")
    spark.sql(f"""create table if not exists `{env}_catalog`.`bronze`.`raw_patient`
              (PatientId VARCHAR(100),
              FirstName VARCHAR(100),
              LastName varchar(100),
              MiddleName varchar(100),
              SSN varchar(100),
              PhoneNumber varchar(100),
              Gender varchar(100),
              DOB varchar(100),
              Address varchar(100),
              ModifiedDate varchar(100),
              Extract_Time varchar(100),
              filename varchar(200));""")
```

```python

create_bronze_table(env)
```
