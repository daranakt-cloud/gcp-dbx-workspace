```python
def remove_duplicates(df):
    print("Removing Duplicates")
    df = df.dropDuplicates()
    print("Duplicates removed")
    return df

def handle_nulls(df,columns):
    print("Handling Nulls in fields")
    df = df.fillna("Unknown", subset=columns)
    return df
```
