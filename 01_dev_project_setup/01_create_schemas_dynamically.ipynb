create a defintion to dynamically create Schema

```python
def create_bronze_schema(path, env):
    print(f"using the catalog {env}_catalog")
    spark.sql(f"use catalog '{env}_catalog'")

    print(f"Creating schema in catalog")          
    spark.sql(f"create schema if not exists `bronze` managed location '{path}/bronze'")
    print("Bronze Schema created")

def create_silver_schema(path, env):
    print(f"using the catalog {env}_catalog")
    spark.sql(f"use catalog '{env}_catalog'")

    print(f"Creating schema in catalog")          
    spark.sql(f"create schema if not exists `silver` managed location '{path}/silver'")
    print("silver Schema created")


def create_gold_schema(path, env):
    print(f"using the catalog {env}_catalog")
    spark.sql(f"use catalog '{env}_catalog'")

    print(f"Creating schema in catalog")          
    spark.sql(f"create schema if not exists `gold` managed location '{path}/gold'")
    print("gold Schema created")
              
```

```python
bronze = spark.sql("DESCRIBE EXTERNAL LOCATION dev_bronze")
silver = spark.sql("DESCRIBE EXTERNAL LOCATION dev_silver")
gold = spark.sql("DESCRIBE EXTERNAL LOCATION dev_gold")

bronze_path = bronze.select("url").collect()[0]["url"]
silver_path = silver.select("url").collect()[0]["url"]
gold_path = gold.select("url").collect()[0]["url"]
```

```python
create_bronze_schema(bronze_path, env)
create_silver_schema(silver_path, env)
create_gold_schema(gold_path, env)
```
